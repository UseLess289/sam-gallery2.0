### Mathématiques des filtres FIR

#### Notations

Let's present the notations we will use : 
- $x(t)$ represents the input signal
- $y(t)$ the output signal
- $h(t)$ is the impulse response of the system
- Any Fourrier transform will be $\mathbf{F}\{\}$ and for any function, $\mathbf F \{f(t)\} = F(e^{j\omega})$
- writing $x[n]$ or $y[n]$ represents a signal sample and we have $x[n] \leftrightarrow X(e^{j\omega})$

### Maths

To calculate $y(t)$ at any time, we must perform a convolution integral : 
$$y(t) = \int_{- \infty}^{+ \infty} h( \tau ) x(t- \tau) d\tau$$

But as you guess, because of causality, we cannot calculate the output based on an input that hasn't arrived yet. Thus,
$$y(t)=\int_0^{+ \infty}h( \tau ) x(t- \tau) d\tau$$

Similary,
$$y[n] = \sum_{k=- \infty}^{+ \infty} h[k] x[n-k]=\sum_{k=0}^{+ \infty} h[k] x[n-k]$$

But we usually use the $B$ vector which contains the $N+1$ $b$ coefficients :

$$t[n]=\sum _{k=0} ^Nb[k]x[n-k]$$


Then, to calculate a value for the current outpout, we must perform the dot product of $B . X$ : 
$$y[0] = B.X = \sum _ {k = 0} ^N b[k]x[0]$$

Now, let's get into a specific case of FIR filters, the notch. If we hvae a 8 element vector $X$, a 4 element vector $B$, we have $8+4-1=11$ element returned. More generally, $$L = len(X) + len(B) - 1$$

In addition, a 4 coeffcients $B$ vector means an order $N=3$. This order corresponds to mowing average filters (low-pass for exemple).
Also, the $h[n]$ coeffs must sum to 1, let's demonstrate that :

We calculate the $Z$-transform for a causal system described as $h[n]$ : 
$$H(z) = \sum_{n=0}^{+\infty} h[n]z^{-n} \Leftrightarrow H(e^{j\omega}) = \sum_{n=0}^{+ \infty} h[n](e^{j \omega})^{-n}$$
Then, to calculate the D.C response of an $N^{th}$ order FIR filter, we set $\omega = 0$ and set the upper limit of summation equals to $N$ :
$$
\begin{align*}
H(e^{j \omega})\lvert_{\omega = 0} &= H(1)\\
    &= \sum_{n=0}^N h[n](1)^{-n}\\
    &= \sum_{n=0}^N h[n]
\end{align*}

Here is why $\sum_{n=0}^N h[n] = 1$.
